```python
from sklearn.datasets import fetch_openml
import numpy as np
import pandas as pd
import pickle
import matplotlib.pyplot as plt
```

# Przygotowanie danych

```python
mnist = fetch_openml('mnist_784', version=1, as_frame=False, parser='auto')
mnist.target = mnist.target.astype(np.uint8)
X = mnist["data"]
y = mnist["target"]
```

# Zadanie 1

```python
from sklearn.cluster import KMeans
```

```python
k_list = [8, 9, 10, 11, 12]
```

```python
k_means_models = []
for k in k_list:
    kmeans = KMeans(n_clusters=k, n_init=10)
    kmeans.fit_predict(X)
    k_means_models.append(kmeans)
```

```python
k_means_models
```

# Zadanie 2

```python
from sklearn.metrics import silhouette_score
```

```python
kmeans_sil = [silhouette_score(X, model.labels_) for model in k_means_models]
```

```python
with open('kmeans_sil.pkl', 'wb') as f:
    pickle.dump(kmeans_sil, f)
```

# Zadanie 4

```python
from sklearn.metrics import confusion_matrix
```

```python
conf_matrix = confusion_matrix(y, k_means_models[2].labels_)
```

```python
print(conf_matrix)
```

# Zadanie 5

```python
indeksy = set()
max_indexes = np.argmax(conf_matrix, axis=1)

for index in max_indexes:
    indeksy.add(index)

max_indexes = list(indeksy)
max_indexes.sort()
```

```python
with open('kmeans_argmax.pkl', 'wb') as f:
    pickle.dump(max_indexes, f)
```

# Zadanie 6

```python
dsts = []

for i in range(300):
    for j in range(X.shape[0]):
        distance = np.linalg.norm(X[i] - X[j])
        if distance != 0:
            dsts.append(distance)
    
```

```python
dsts.sort()
```

```python
dsts[:10]
```

```python
with open('dist.pkl', 'wb') as f:
    pickle.dump(dsts[:10], f)
```

# Zadanie 7

```python
from sklearn.cluster import DBSCAN
```

```python
ls = np.array(dsts[:3])
```

```python
s = ls.mean()
```

```python
epsilon = s
models = []

while(epsilon <= 1.1*s):
    model = DBSCAN(eps=epsilon)
    model.fit(X)
    models.append(model)
    epsilon += 0.04*s
```

```python
lengths = []
for i in range(len(models)):
    lengths.append(len(np.unique(models[i].labels_)))
```

```python
with open('dbscan_len.pkl', 'wb') as f:
    pickle.dump(lengths, f)
```
